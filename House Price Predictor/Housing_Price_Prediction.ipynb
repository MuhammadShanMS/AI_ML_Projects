This notebook predicts house prices using Linear Regression. Markdown cells explain each step, and comments explain what the code is doing. The original code logic has been kept unchanged.

## 1. Load the Dataset
First, we import Pandas and load the Housing CSV file into a DataFrame called `data`.


```python
# Load the Housing dataset into a Pandas DataFrame
import pandas as pd
data = pd.read_csv("Housing.csv")
```

## 2. View the First Rows
`head()` shows the first 5 rows so we can understand what the dataset looks like.


```python
# Display the first 5 rows
data.head()
```

## 3. Check Dataset Size
`shape` tells us the number of rows and columns in the dataset.


```python
# Check number of rows and columns
data.shape
```

## 4. Check Dataset Information
`info()` shows column names, non-null values, and data types.


```python
# Check column names, data types, and non-null counts
data.info()
```

## 5. Check Missing Values
We check whether any column contains missing (`NaN`) values.


```python
# Count missing values in each column
data.isnull().sum()
```

## 6. Check Duplicate Rows
We check whether the dataset contains duplicate rows.


```python
# Count duplicate rows
data.duplicated().sum()
```

## 7. Convert Yes/No to 1/0
Machine-learning models work with numbers, so the binary `yes/no` columns are converted to `1/0`.


```python
# Convert binary yes/no columns into numeric 1/0 values# This makes these categorical values usable by the ML model.
data["mainroad"] = data["mainroad"].map({"yes": 1, "no": 0})
data["guestroom"] = data["guestroom"].map({"yes": 1, "no": 0})
data["basement"] = data["basement"].map({"yes": 1, "no": 0})
data["hotwaterheating"] = data["hotwaterheating"].map({"yes": 1, "no": 0})
data["airconditioning"] = data["airconditioning"].map({"yes": 1, "no": 0})
data["prefarea"] = data["prefarea"].map({"yes": 1, "no": 0})
```

## 8. Verify the Conversion
We display the data again to confirm that the `yes/no` values were converted successfully.


```python
# Verify that the conversions were applied
data.head()
```

## 9. Check Data Types Again
We check the data types after converting the binary columns.


```python
# Check the updated data types
data.dtypes
```

## 10. Find Categorical Columns
We identify the remaining text/categorical columns that still need encoding.


```python
# Find the remaining string/categorical columns
cat_col = data.select_dtypes(include=["str"])
```

## 11. View the Categorical Columns
This confirms which columns were identified as categorical.


```python
# Display the categorical columns
cat_col
```

## 12. Create a One-Hot Encoder
`OneHotEncoder` will convert categorical values such as `furnished`, `semi-furnished`, and `unfurnished` into numeric columns.


```python
# Create an encoder for categorical data
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
```

## 13. Apply One-Hot Encoding
`fit_transform()` learns the categories and converts them into numerical values.


```python
# Learn the categories and convert them to numeric values
encoded_cat = encoder.fit_transform(cat_col)
```

## 14. View the Encoded Values
We inspect the numerical output produced by the encoder.


```python
# Display the encoded values
encoded_cat
```

## 15. Get Encoded Column Names
We get the names of the new columns created by One-Hot Encoding.


```python
# Get the names of the newly created encoded columns
encoder.get_feature_names_out()
```

## 16. Convert Encoded Data to a DataFrame
The encoded array is converted into a Pandas DataFrame and given meaningful column names.


```python
# Convert the encoded array into a DataFrame
encoded_cat = pd.DataFrame(encoded_cat,columns=encoder.get_feature_names_out())
```

## 17. Check the Encoded Data
We display the first rows of the encoded DataFrame to verify it.


```python
# Check the encoded DataFrame
encoded_cat.head()
```

## 18. Remove the Original Categorical Column
The original text column is removed because its encoded version is already available.


```python
# Remove the original categorical column
numeric_data = data.drop(columns=cat_col.columns)
```

## 19. Combine the Data
We combine the numeric data and the newly encoded categorical data into one final dataset.


```python
# Combine numeric and encoded columns side-by-side
final_data = pd.concat([numeric_data.reset_index(drop=True),encoded_cat.reset_index(drop=True)],axis=1)
```

## 20. View the Final Dataset
We check the processed dataset before building the machine-learning model.


```python
# Check the final processed dataset
final_data.head()
```

## 21. Check the Final Dataset Shape
This should ideally use `final_data.shape` because `final_data` is the processed dataset after encoding. The current code checks `data.shape` instead, which is a small mistake.


```python
# NOTE: This checks the original data, not final_data.
data.shape
```

## 22. Separate Features (X) and Target (y)
`X` contains the information used to predict the price. `y` contains the actual house price that we want to predict.


```python
# X = input features; y = target price
X = final_data.drop(columns="price")
y = final_data["price"]
```

## 23. Split Data into Training and Testing Sets
70% of the data is used for training and 30% is kept for testing. The model does not see the test data during training.


```python
# Use 70% for training and 30% for testing
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

## 24. Create the Linear Regression Model
We create a Linear Regression model. At this point, the model has not learned from the data yet.


```python
# Create the Linear Regression model
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
```

## 25. Train the Model
`fit()` trains the model using the training features and their actual prices.


```python
# Train the model using the training data
lr.fit(X_train,y_train)
```

## 26. Make Predictions
The trained model uses the unseen test features to predict house prices.


```python
# Predict prices for the unseen test data
y_pred = lr.predict(X_test)
```

## 27. Import Evaluation Metrics
We import MAE, MSE, and R² to measure model performance. NumPy is used to calculate RMSE.


```python
# Import metrics used to evaluate the model
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
```

## 28. Calculate MAE
MAE tells us the average absolute difference between the actual prices and predicted prices. Lower is better.


```python
# Mean Absolute Error: average absolute prediction error
mae = mean_absolute_error(y_test, y_pred)
mae
```

## 29. Calculate MSE
MSE squares the prediction errors before averaging them. Lower is better, but the value is harder to interpret because the errors are squared.


```python
# Mean Squared Error: average squared prediction error
mse = mean_squared_error(y_test, y_pred)
mse
```

## 30. Calculate RMSE
RMSE is the square root of MSE. It is in the same unit as the house price, making it easier to interpret. Lower is better.


```python
# Root Mean Squared Error: square root of MSE
rmse = np.sqrt(mse)
rmse
```

## 31. Calculate R² Score
R² shows how well the model explains variation in house prices. A value closer to 1 generally indicates a better fit.


```python
# R-squared: how well the model explains variation in the target
r2 = r2_score(y_test, y_pred)
r2
```

## 32. Compare Actual and Predicted Prices
We create a table containing the actual test prices and the prices predicted by the model.


```python
# Put actual and predicted prices together for easy comparison
model_comparison = pd.DataFrame({"Actual": y_test.reset_index(drop=True), "Predicted": y_pred})
```
